# PySpark
The easiest way to use Spark is through Databricks, as they have tools to simplify the process. However, it is possible to run Spark locally.
* Install Java Development Kit (JDK) 
  * For Apache Spark 3.3 - 3.5, you need JDK 8, 11, or 17
  * For Apache Spark 4.x, you need JDK 17, 21, or 25.
* Install Apache Spark 
* Install Python and Jupyter Lab
* Launch Jupyter Lab

```python
# Set the PySpark environment variables
import os
os.environ['SPARK_HOME'] = "/Users/michael.olson/Apps/Spark"
os.environ['PYSPARK_DRIVER_PYTHON'] = 'jupyter'
os.environ['PYSPARK_DRIVER_PYTHON_OPTS'] = 'lab'
os.environ['PYSPARK_PYTHON'] = 'python'
```

```python
# Import PySpark
from pyspark.sql import SparkSession
```

```python
# Create a SparkSession
spark = SparkSession.builder \
     .appname('PySpark-Get-Started') \
     .getOrCreate()
```

```python
# Test the setup
data = [('Alice', 25),
        ('Bob', 30),
        ('Charlie', 35)]

df = spark.createDataFrame(data, ['Name', 'Age'])

df.show()
```

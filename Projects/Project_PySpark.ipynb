## Overview

PySpark is the Python interface to Apache Spark, a framework for processing large datasets across many machines at once. The syntax will feel familiar if you know pandas — but instead of a `DataFrame` that lives entirely in your computer's memory, a Spark DataFrame is built to scale up to datasets far bigger than a single machine can hold.

This project is a **gentle first look** at PySpark using a tiny dataset, just to get comfortable with the basic building blocks: starting a Spark session, creating a DataFrame, and running a few simple queries.

## Learning objectives

By the end of this project you should be able to:
- Start and stop a `SparkSession`
- Create a Spark DataFrame from Python data
- View, select, filter, and sort rows in a Spark DataFrame
- Group and aggregate data with `groupBy()`

## Part 1: Start a Spark session

Everything in PySpark starts with a `SparkSession` — it's your connection to the Spark engine. `.master("local[*]")` tells Spark to run on your own machine, using all available cores, rather than a cluster.

```
spark = SparkSession.builder \
    .appName("MiniProject") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")  # keep the output tidy

print("Spark session started.")
```

## Part 2: Create a DataFrame

Run the cell below to build a small Spark DataFrame of movies, with a `title`, `genre`, `rating`, and release `year` for each one. (No changes needed here.)

```
data = [
    ("Inception",           "Sci-Fi",  8.8, 2010),
    ("The Notebook",        "Romance", 7.8, 2004),
    ("Interstellar",        "Sci-Fi",  8.6, 2014),
    ("La La Land",          "Romance", 8.0, 2016),
    ("The Dark Knight",     "Action",  9.0, 2008),
    ("Mad Max: Fury Road",  "Action",  8.1, 2015),
    ("Arrival",             "Sci-Fi",  7.9, 2016),
    ("Pride & Prejudice",   "Romance", 7.8, 2005),
]
columns = ["title", "genre", "rating", "year"]

movies_df = spark.createDataFrame(data, columns)
movies_df.show()
```
